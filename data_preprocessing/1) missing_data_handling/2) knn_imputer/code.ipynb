## KNN Imputer: Small explanation is in the file 'all_about_it.md' and mostly explained here with code executions
KNN imputer would be recommended when features are dependent (Correlated with each other). Preserves relationships between features (e.g., higher Experience should correspond to higher Salary).
- Higher accuracy because it tailors values based on similar instances.
- Robust to outliers because local neighbors dictate the value, not global statistics.

But also comes with some cons
- when it comes to accuracy, it's Slower $O(N^2)$ because it computes distance matrices across all rows.
- Best for small to medium datasets. Can be very slow or memory-intensive on millions of rows.

### Things to know related to Knn imputer:
#### What is the $k$ Value and How to Choose It?

In KNNImputer, $k$ represents the number of nearest neighbors (similar rows) the algorithm looks at to calculate the average for a missing value.

- If $k=1$, it copies the missing value directly from the single most similar row.
- If $k=5$ (the default in scikit-learn), it finds the $5$ most similar rows, averages their values, and fills the gap.

#### How to Choose the Best $k$:
- Small $k$ (e.g., 1–3): Captures local patterns well, but can be sensitive to noise or outliers in your data.
- Large $k$ (e.g., 15–20+): Smooths out noise, but if $k$ is too large, it starts averaging across dissimilar rows, acting almost like SimpleImputer.
- General Rule of Thumb: A common starting choice for small-to-medium datasets is $k = \sqrt{N}$ (where $N$ is the total number of rows) or picking an odd number between 3 and 10 to prevent tie-breaking issues.
- Best Practice: Use GridSearchCV with cross-validation in your ML pipeline to test values (e.g., $k \in [3, 5, 7, 9]$) and see which yields the highest downstream model performance.

```python
# Example dataset

import numpy as np
import pandas as pd

# Create a sample dataset with correlated features and missing values
data = {
    'Experience_Years': [1, 2, 3, 10, 11, 12, 20, 22],
    'Age': [22, 24, 25, 33, 35, 36, 48, 52],
    'Salary': [30000, 35000, 38000, 85000, np.nan, 95000, 150000, 160000] # Missing senior-level salary
}

df = pd.DataFrame(data)

print("=== ORIGINAL DATASET WITH MISSING VALUES ===")
print(df)
```

```python
#Knn imputer code
# instead of copy, we can also replace the original column with the transformed values if wanted.

from sklearn.impute import KNNImputer

# Create a copy so original df stays untouched
df_knn = df.copy()

# Initialize KNNImputer looking at 3 nearest neighbors (k=3)
knn_imp = KNNImputer(n_neighbors=3)

# Fit and transform across all correlated columns
df_knn_filled = pd.DataFrame(knn_imp.fit_transform(df_knn), columns=df_knn.columns)

# to get the original column replaced, you can change the line like: df_knn = pd.DataFrame(knn_imp.fit_transform(df_knn), columns=df_knn.columns) and also delete that copy line above.

print("=== KNN IMPUTER RESULT (k=3) ===")
print(df_knn_filled)
```
