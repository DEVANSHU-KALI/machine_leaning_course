## KNN Imputer: Small explanation is in the file 'all_about_it.md' and mostly explained here with code executions
KNN imputer would be recommended when features are dependent (Correlated with each other). Preserves relationships between features (e.g., higher Experience should correspond to higher Salary).
- Higher accuracy because it tailors values based on similar instances.
- Robust to outliers because local neighbors dictate the value, not global statistics.

But also comes with some cons
- when it comes to accuracy, it's Slower $O(N^2)$ because it computes distance matrices across all rows.
- Best for small to medium datasets. Can be very slow or memory-intensive on millions of rows.
